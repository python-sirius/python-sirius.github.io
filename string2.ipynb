она должна возвращать.

```python
def test(got, expected):
    if got == expected:
        prefix = ' OK '
    else:
        prefix = '  X '
    print('%s Получено: %s | Ожидалось: %s' % 
          (prefix, repr(got), repr(expected)))
```
